```python
import os
import requests

# ============================================================
# 1. CONFIGURATION (self-contained)
# ============================================================

# Output directory
DEM_DIR = "./Data/Datasets/ground_truth-9CH/Raw/DEM"
os.makedirs(DEM_DIR, exist_ok=True)

# Your bounding boxes (min_lon, min_lat, max_lon, max_lat)
TP_BOUNDING_BOXES = [
                        (-1.875,51.145,-1.775,51.205),
                        (-1.895,51.405,-1.815,51.455),
                        (-1.895,51.31,-1.85,51.338),
                        (-1.12,51.335,-1.05,51.38),
                        (1.255,52.565,1.325,52.6),
                        (1.275,52.608,1.322,52.635),
                        (-2.445,50.7,-2.407,50.722),
                        (1.246,52.565,1.293,52.592),
                        (-1.82,51.165,-1.72,51.225),
                        (-1.84,51.425,-1.76,51.475),
                        (-1.84,51.33,-1.795,51.358),
                        (-1.688,51.098,-1.644,51.128),
                        (-1.065,51.355,-0.995,51.4),
                        (1.31,52.585,1.38,52.62),
                        (1.33,52.628,1.377,52.655),
                        (-2.39,50.72,-2.352,50.742),
                        (1.301,52.585,1.348,52.612),
]

# OpenTopography API key
API_KEY = "fa950cf7b479a07beeea17b36be91b56"

# OpenTopography DEM endpoint
OPENTOPO_URL = (
    "https://portal.opentopography.org/API/globaldem?"
    "demtype=COP30&south={south}&north={north}&west={west}&east={east}"
    "&outputFormat=GTiff&API_Key={key}"
)


# ============================================================
# 2. Download helper
# ============================================================
def download(url, out_path):
    print(f"[DOWNLOAD] {url}")

    r = requests.get(url, stream=True)
    if r.status_code != 200:
        print(f"[ERROR] {r.status_code} {url}")
        return

    with open(out_path, "wb") as f:
        for chunk in r.iter_content(8192):
            f.write(chunk)

    print(f"[OK] Saved {out_path}")


# ============================================================
# 3. MAIN LOOP — iterate over bounding boxes
# ============================================================
for i, (west, south, east, north) in enumerate(TP_BOUNDING_BOXES):
    print(f"\n=== Processing DEM tile {i}: {west, south, east, north} ===")

    url = OPENTOPO_URL.format(south=south,
                              north=north,
                              west=west,
                              east=east,
                              key=API_KEY
    )
    tile_dir = os.path.join(DEM_DIR, f"tile {i}")
    os.makedirs(tile_dir, exist_ok=True)
    out_path = os.path.join(tile_dir, f"dem_tile_{i}.tif")
    download(url, out_path)
```
